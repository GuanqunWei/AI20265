## 2.1 产生式

1943 年，美国数学家波斯特（E. Post）首先提出“产生式”。1972 年，纽厄尔和西蒙在研究人类认知模型时开发了基于规则的产生式系统。产生式通常用于表示事实、规则以及它们的不确定性度量，适合表示事实性知识和规则性知识。

### 1. 确定性规则知识的产生式表示

基本形式：

```text
IF P THEN Q
```

或者：

```text
P -> Q
```

例如：

```text
r4: IF 动物会飞 AND 会下蛋 THEN 该动物是鸟
```

### 2. 不确定性规则知识的产生式表示

基本形式：

```text
IF P THEN Q （置信度）
```

例如：

```text
IF 发烧 THEN 感冒 (0.6)
```

### 3. 确定性事实性知识的产生式表示

可以使用三元组表示：

```text
（对象，属性，值）
```

或者：

```text
（关系，对象1，对象2）
```

例如：

```text
老李年龄是 40 岁：（Li, age, 40）
老李和老王是朋友：（friend, Li, Wang）
```

### 4. 不确定性事实性知识的产生式表示

可以使用四元组表示：

```text
（对象，属性，值，置信度）
```

或者：

```text
（关系，对象1，对象2，置信度）
```

例如：

```text
老李年龄很可能是 40 岁：（Li, age, 40, 0.8）
老李和老王不大可能是朋友：（friend, Li, Wang, 0.1）
```

### 产生式与谓词逻辑中蕴含式的区别

1. 除逻辑蕴含外，产生式还可以包括各种操作、规则、变换、算子、函数等。例如，“如果炉温超过上限，则立即关闭风门”是一个产生式，但不只是逻辑蕴含式。
2. 蕴含式通常只能表示精确知识，而产生式既可以表示精确知识，也可以表示不精确知识。产生式匹配可以是精确匹配，也可以是不精确匹配，只要相似度满足指定条件即可。

### 产生式的 BNF 形式描述

```text
<产生式> ::= <前提> <结论>
<前提> ::= <简单条件> | <复合条件>
<结论> ::= <事实> | <操作>
<复合条件> ::= <简单条件> AND <简单条件> [AND <简单条件> ...]
             | <简单条件> OR <简单条件> [OR <简单条件> ...]
<操作> ::= <操作名> [(<变元>, ...)]
```

## 2.2 产生式系统

产生式系统一般由三部分组成：

1. **规则库**：用于描述相应领域内知识的产生式集合。
2. **综合数据库**：也称事实库、上下文或黑板，用于存放问题求解过程中的当前信息。
3. **控制系统**：也称推理机构，由一组程序组成，负责整个产生式系统的运行，实现问题求解。

控制系统的主要工作包括：

1. 从规则库中选择规则，与综合数据库中的已知事实进行匹配。
2. 当多条规则匹配成功时，进行冲突消解。
3. 执行某一条规则。如果规则右部是结论，则把结论加入综合数据库；如果右部是操作，则执行对应操作。
4. 对于不确定性知识，在执行规则时还需要计算结论的不确定性。
5. 检查综合数据库中是否包含最终结论，决定是否停止系统运行。

## 2.4 产生式表示法的特点

### 优点

1. **自然性**：规则形式接近人类表达知识的方式。
2. **模块性**：每条规则相对独立，便于增加、删除和修改。
3. **有效性**：适合表示经验性知识和启发式规则。
4. **清晰性**：规则结构明确，便于解释推理过程。

### 缺点

1. **效率不高**：规则数量较多时，匹配过程可能耗时。
2. **不能很好表达结构性知识**：对于具有复杂层次结构或关系结构的知识，表达能力有限。

### 适合产生式表示的知识

1. 领域知识之间关系不密切，不存在复杂结构关系。
2. 经验性和不确定性知识，且相关领域中没有严格、统一的理论。
3. 问题求解过程可以表示为一系列相对独立的操作，每个操作可以表示为一条或多条产生式规则。
